# 딕셔너리 연습


딕셔너리는 **'이름표(key)'를 이용해서 값(value) 을 찾는 자료형**입니다.  
예를 들어 학생 정보에서:

- `'name'` → `'kim'`
- `'phone'` → `'010-9999-1234'`

처럼 **하나의 key에 하나의 value가 연결**됩니다.

---

## 이번 시간에 배우는 것
1. 딕셔너리의 기본 구조  
2. key를 이용해 value 꺼내기  
3. 중첩 딕셔너리 접근하기  
4. API 데이터처럼 복잡한 구조 읽기  
5. 딕셔너리 + 리스트 구조를 DataFrame으로 바꾸기


````python
import pandas as pd
````

## 1. 딕셔너리란?

딕셔너리는 `{ }` 중괄호로 만들며,  
각 요소는 `key : value` 형태로 작성합니다.

형식은 다음과 같습니다.

```python
변수명 = {'key1': value1, 'key2': value2}
```

예를 들어 학생 정보를 딕셔너리로 만들 수 있습니다.


````python
student = {'name': 'kim', 'phone': '010-9999-1234', 'date': 1118}
student
````

위 결과를 보면 딕셔너리는 다음처럼 구성됩니다.

- `'name'` 는 key
- `'kim'` 은 value
- `'phone'` 는 key
- `'010-9999-1234'` 는 value

즉, **key를 이용해서 원하는 값을 찾을 수 있는 구조**입니다.


## 2. key로 value 꺼내기

딕셔너리에서 값을 꺼낼 때는 대괄호 `[ ]` 안에 key를 넣습니다.

형식:

```python
변수명['key']
```


### 2-1. student에서 `'name'` 값 꺼내기

````python
student['name']
````

`student['name']` 은  
student 딕셔너리에서 `'name'` 이라는 key에 연결된 값을 가져오라는 뜻입니다.


### 2-2. student에서 `'phone'` 값 꺼내기

````python
student['phone']
````

### 2-3. student에서 `'date'` 값 꺼내기

````python
student['date']
````

## 3. 딕셔너리에서 자주 하는 실수

### 실수 1: key 이름을 정확히 쓰지 않음
예를 들어 `'name'` 이 아니라 `'Name'` 이라고 쓰면 오류가 납니다.  
파이썬은 **대소문자를 구분**합니다.

### 실수 2: 대괄호 대신 괄호를 사용함
```python
student('name')   # 잘못된 예
student['name']   # 올바른 예
```

### 실수 3: key에 따옴표를 빼먹음
문자 key는 보통 `'name'` 처럼 따옴표로 써야 합니다.


## 4. API 데이터처럼 더 복잡한 구조 보기

실제 데이터 분석에서는 단순한 딕셔너리보다  
**딕셔너리 안에 또 딕셔너리**, 또는  
**딕셔너리 안에 리스트**가 들어 있는 구조를 많이 만납니다.

특히 **Open API / JSON 데이터**가 이런 형태로 자주 제공됩니다.


````python
data = {
    'response': {
        'body': {
            'totalCount': 545,
            'items': [
                {'sn': '534', 'dataDate': '0416', 'issueVal': '157'},
                {'sn': '538', 'dataDate': '04-17', 'issueVal': '201'}
            ],
            'pageNo': 1,
            'numOfRows': 100
        },
        'header': {
            'resultMsg': 'NORMAL_CODE',
            'resultCode': '00'
        }
    }
}
````

````python
data
````

위 구조를 말로 설명하면 다음과 같습니다.

- 가장 바깥쪽은 딕셔너리
- 그 안에 `'response'` 라는 key가 있음
- `'response'` 의 값도 딕셔너리
- 그 안에 `'body'`, `'header'` 가 있음
- `'body'` 안의 `'items'` 는 **리스트**
- 그 리스트 안에는 **딕셔너리 여러 개**가 들어 있음


## 5. 바깥쪽 key부터 하나씩 접근하기

복잡한 구조는 한 번에 보려 하지 말고  
**바깥쪽부터 차례대로** 접근하면 쉽습니다.


### 5-1. `data` 에서 `'response'` 값 꺼내기

````python
data['response']
````

`data['response']` 의 결과를 보면 또 딕셔너리입니다.  
즉, 한 번 꺼냈더니 끝이 아니라 **그 안에 또 key들이 들어 있다**는 뜻입니다.


### 5-2. `'response'` 안에서 다시 `'body'` 값 꺼내기

````python
data['response']['body']
````

이 코드는 아래 순서로 읽으면 됩니다.

1. `data['response']` 로 `'response'` 값을 꺼낸다  
2. 그 결과에서 다시 `['body']` 를 붙여 `'body'` 값을 꺼낸다  

즉, **안쪽으로 한 단계 더 들어간 것**입니다.


### 5-3. `'body'` 안의 `'totalCount'` 값 꺼내기

````python
data['response']['body']['totalCount']
````

### 5-4. `'body'` 안의 `'items'` 값 꺼내기

````python
data['response']['body']['items']
````

`items` 의 결과는 **리스트(list)** 입니다.  
그리고 그 리스트 안에는 각 관측값이 **딕셔너리 형태**로 들어 있습니다.

즉:

- `items` 전체 = 여러 행(row)의 묶음
- 각 행 = 딕셔너리 1개


## 6. DataFrame으로 바꾸기

`items` 는 **딕셔너리들의 리스트** 형태이므로  
`pd.DataFrame()` 으로 데이터프레임으로 바꿀 수 있습니다.


````python
df = pd.DataFrame(data['response']['body']['items'])
df
````

**API 데이터를 분석 가능한 데이터프레임으로 바꾸는 단계**

## 8. 정리
이번 시간의 핵심은 다음 3가지입니다.

### 1) 딕셔너리는 `key : value` 구조이다
예:
```python
student['name']
```

### 2) 중첩된 딕셔너리는 차례대로 들어간다
예:
```python

```

### 3) 딕셔너리들의 리스트는 DataFrame으로 바꿀 수 있다
예:
```python
pd.DataFrame(data['response']['body']['items'])
```
