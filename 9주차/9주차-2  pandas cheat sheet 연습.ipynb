## DataFrame 만들기

**DataFrame**은 pandas의 핵심 자료구조로, **2차원 표(테이블)** 형태입니다.

| 용어 | 설명 | 예시 |
|------|------|------|
| **행 (row)** | 가로줄, 하나의 관측값(데이터) | 한 학생의 성적 정보 |
| **열 (column)** | 세로줄, 데이터의 속성(변수) | 이름, 국어, 영어, 수학 |
| **인덱스 (index)** | 각 행의 식별자 | 기본값: 0, 1, 2, ... |

DataFrame을 만드는 방법은 **입력 데이터의 형태**에 따라 다양합니다. 아래 예제들을 직접 따라 쓰며 확인해 보세요!

### 다음 코드를 보고 아래에 다시 쓰고 실행하며 확인하세요!

````python
df=pd.DataFrame([4,5,6],columns=['A'])       
df
````

> **방법 1 — 1차원 리스트로 만들기**
> `[4, 5, 6]`처럼 1차원 리스트를 넣으면 **1개의 열**이 생성됩니다.       
> `columns=['A']`로 열 이름을 지정하며, 인덱스는 자동으로 **0, 1, 2**가 됩니다.

````python
df=pd.DataFrame([4,5,6],columns=['A'])       
df
````

````python
df=pd.DataFrame([[4,7,10]])
df
````

> **방법 2 — 2차원 리스트 (열 이름 없음)**
> `[[4, 7, 10]]`처럼 리스트 안에 리스트를 넣으면 **1개의 행, 3개의 열**이 생성됩니다.         
> 열 이름을 지정하지 않으면 자동으로 **0, 1, 2** 숫자가 열 이름이 됩니다.

````python
df=pd.DataFrame([[4,7,10]])
df
````

````python
df=pd.DataFrame([[4,7,10]],columns=['A', 'B','C'])
df
````

> **방법 3 — 2차원 리스트 + 열 이름 지정**
> `columns=['A', 'B', 'C']`로 열 이름을 명시적으로 지정합니다.

````python
df=pd.DataFrame([[4,7,10]],columns=['A', 'B','C'])
df
````

````python
df=pd.DataFrame({'A': [4,5,6],
                 'B': [7,8,9],
                 'C': [10,11,12]})
df
````

> **방법 4 — 딕셔너리(열 단위)로 만들기**    
> `{'열이름': [값들의 리스트]}` 형태입니다.        
> **키(key) = 열 이름**, **값(value) = 해당 열의 데이터 리스트**      
> 각 열을 독립적으로 정의할 수 있어 직관적입니다.

df=pd.DataFrame({'A': [4,5,6],
                 'B': [7,8,9],
                 'C': [10,11,12]})
df

````python
df=pd.DataFrame([{'A': 4,
                  'B': 7,
                  'C': 10}])
df
````

> **방법 5 — 딕셔너리 1개를 리스트로 감싸기 (1행)**
> 딕셔너리 하나를 리스트로 감싸면 **1개의 행**이 만들어집니다.      
> 딕셔너리의 키 → 열 이름, 딕셔너리의 값 → 해당 셀의 데이터     

````python
df=pd.DataFrame([{'A': 4,
                  'B': 7,
                  'C': 10}])
df
````

````python
df=pd.DataFrame([{'A': 4, 'B': 7,'C': 10},
                {'A': 5, 'B': 8,'C': 11},
                {'A': 6, 'B': 9,'C': 12}])
df
````

> **방법 6 — 딕셔너리 리스트로 만들기 (가장 실용적!)**
> **딕셔너리 1개 = 행(row) 1개** 입니다.
> API 응답, JSON 파일 등 실제 데이터는 대부분 이 형태로 되어 있어, 실무에서 가장 많이 사용됩니다.        
> ```
> [{'A':4,'B':7,'C':10},   ← 0번 행
>  {'A':5,'B':8,'C':11},   ← 1번 행
>  {'A':6,'B':9,'C':12}]   ← 2번 행
> ```

````python
df=pd.DataFrame([{'A': 4, 'B': 7,'C': 10},
                {'A': 5, 'B': 8,'C': 11},
                {'A': 6, 'B': 9,'C': 12}])
df
````

### 여러 열 선택


여러 열을 동시에 선택할 때는 **대괄호 두 겹** `df[['열1', '열2']]`을 사용합니다.

```
df[['A', 'B']]
   ↑        ↑
   바깥 []   안쪽 []는 열 이름 리스트
```

- 안쪽 `[]`: 선택할 열 이름들의 **리스트**
- 바깥쪽 `[]`: DataFrame **인덱싱** 연산자
- 결과는 항상 **DataFrame** 형태로 반환됩니다!

````python
df[['A','B']]
````

````python
df[['A','B']]
````

### 하나의 행 선택

### 행(Row) 선택: loc 사용법

`df.loc[인덱스]`로 특정 행을 선택합니다.

- `loc`은 **라벨(label) 기반** 인덱싱입니다
- 기본 인덱스가 0, 1, 2...일 때 해당 숫자로 접근
- 결과는 **Series** 형태로 반환됩니다 (인덱스 = 열 이름, 값 = 해당 행의 데이터)

> 예: `df.loc[0]` → 0번 행의 데이터를 Series로 반환

````python
df.loc[0]
````

````python
df.iloc[0]
````

### 여러 행 선택


`df.loc[[인덱스1, 인덱스2]]`처럼 **리스트**로 여러 행을 선택합니다.

- 결과는 **DataFrame** 형태로 반환됩니다
- 연속된 행이 아니어도 됩니다! (예: 0번 행과 2번 행만 선택 가능)

````python
df.loc[[0,2]]
````

````python
df.loc[[0,2]]
````

````python
df.iloc[[0,2]]
````

````python
df.iloc[[0,2]]
````

### loc 사용하여 행 , 열 선택


`df.loc[행 선택, 열 선택]` 형태로 **행과 열을 동시에** 지정할 수 있습니다.

| 코드 | 의미 |
|------|------|
| `df.loc[:, 'A']` | **모든 행**, A열만 선택 |
| `df.loc[:, ['A','C']]` | **모든 행**, A열과 C열 선택 |
| `df.loc[1, ['A','C']]` | **1번 행**, A열과 C열 선택 |
| `df.loc[[0,1], ['A','C']]` | **0,1번 행**, A열과 C열 선택 |

>  `:` 는 "전체(all)"를 의미합니다 — 파이썬 슬라이싱과 동일한 개념!

````python
df.loc[:, 'A']                                            # df.loc[:, ['A']] 
````

````python
df.loc[:, 'A']   
````

````python
df.loc[:, ['A','C']]
````

````python
df.loc[:, ['A','C']]
````

````python
df.loc[1, ['A','C']]
````

````python
df.loc[1, ['A','C']]
````

````python
df.loc[[0 ,1], ['A','C']]
````

````python
df.loc[[0 ,1], ['A','C']]
````

### 조건을 만족하는 데이터 (하나의 데이터는 하나의 행) 선택 
#### 불리언(Boolean) 인덱싱 — 조건에 맞는 행만 필터링!

조건을 이용해 원하는 데이터만 골라내는 강력한 기능입니다.

**동작 원리 (2단계):**

1. **조건식** `df['A'] > 5` → 각 행에 대해 True / False 반환
2. **필터링** `df[조건]` → True인 행만 선택하여 반환

**여러 조건 조합:**
```python
df[(df['A'] > 4) & (df['B'] < 9)]   # AND 조건 (둘 다 만족)
df[(df['A'] < 5) | (df['C'] > 11)]  # OR 조건 (하나만 만족해도)
```

> 여러 조건을 쓸 때는 각 조건을 **소괄호 `()`로 반드시 감싸야** 합니다!

````python
df['A']>5
````

````python
df[df['A']>5]
````

````python
df.loc[df['A']>5]
````

````python
df.loc[df['A']>5]
````

````python
df.loc[df['A']>5, ['A','B']]
````

````python
df.loc[df['A']>5, ['A','B']]
````
