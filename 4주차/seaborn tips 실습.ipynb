그래프 해석

1. 비대칭적 분포 (오른쪽으로 꼬리)
 - 대부분의 팁은 2~4달러 사이에 몰려 있음
 - 10달러 정도의 팁은 거의 없음
 - 오른쪽으로 긴 꼬리(long right tail) 를 가진 우측 비대칭 분포 (right-skewed distribution) (즉 항상 정규분포가 아님)

2. 최빈값 (가장 많은 값)
 - 약 2달러 전후의 구간에 가장 많은 사람들이 팁을 주었음
 - 2달러팁을 준 경우가 50건 이상으로 가장 많음

3. 이상치 존재 가능
 - 6달러 이상 팁은 아주 드물게 발생
 - 이런 값은 이상치(outlier) 일 수도 있고, 아닐 수도 있는데 모델링이나 평균 계산 시 주의가 필요

4. 중앙값 vs 평균
 - 평균은 오른쪽 꼬리에 의해 끌려 올라갈 수 있음
 - 따라서 중앙값 < 평균일 가능성이 높음
 - 평균 대신 중앙값이 데이터를 대표하는 값이 되는 사례


해석 예: 대부분 손님은 2~4달러 사이의 팁을 주며, 데이터는 오른쪽으로 비대칭적이고 이상치도 존재함!

### 6. scatterplot: 결제금액 vs 팁

```python
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')
```

```python
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')
```

그래프 해석

- 결제 금액이 클수록 팁도 증가하는 경향, 즉 더 많이 결제할수록 팁도 많이 주는 경향이 있음

- 전체적으로 우상향하는 패턴이므로 양의 상관관계가 있음

- 성별 차이 : 

    - 남성과 여성 모두 유사한 패턴을 보임

    - 고액 팁(6~10달러)은 남성에게서 많이 관측됨

    - 여성은 중간 정도의 결제금액 (~30달러)까지 팁 분포가 더 좁은 편

- 고액 결제와 고액 팁은 드문 조합이라 이상치 또는 VIP일 수 있음

violinplot 구성요소	

- 중심 흰 점 :	중앙값 (Median)

- 굵은 검정 막대	: IQR (Interquartile Range) = Q1~Q3

- 전체 실루엣(모양) : 분포의 밀도 (KDE) — 부풀수록 데이터가 많다는 뜻

<h3 align="center"> 요일별 결제금액 분포 비교 </h3> 


| 요일 | 중앙값 | 분포 형태 | 최대 결제금액 | 특이사항 |
|:--------:|:--------:|:--------:|:--------:|:--------:|
| Thur	|약 16달러	|좌우 대칭, 분산 적음	|40달러	|중간값 낮고 안정적
| Fri	|약 16달러	|분포가 위쪽으로 더 퍼짐	|45달러	|이상치 없음, 분포 넓음
| Sat	|약 18달러	|양쪽 날개 넓음|	60달러 이상|	가장 큰 분산과 이상치
| Sun	|약 17달러	|좌우 균형 잡힌 종형|	50달러|	중간값 높고 분포 넓음

해석 예

토요일(Sat): 가장 넓고 고액 결제도 많음 즉 가장 비싼 손님 많음

금/일요일(Fri/Sun): 결제금액 중간~상단으로 집중 따라서 중상위 결제 많음

목요일(Thur): 가장 집중된 분포로 예측 안정성 좋음

### 9. 상관계수 구하기와  heatmap

<h3 align="center"> 상관계수(correlation coefficient) </h3> 

|값 범위 | 의미|
|:--------:|:--------:|  
|+1에 가까울수록 | 강한 양의 상관관계 (같이 증가)|
|0에 가까울수록 | 거의 무관|
|-1에 가까울수록 | 강한 음의 상관관계 (한쪽이 증가할 때 다른 쪽은 감소)|

```python
df[['total_bill', 'tip', 'size']].corr()
```

```python
df[['total_bill', 'tip', 'size']].corr()
```

### 수치값으로 상관계수를 나타낼 수도 있으나 데이터의 크기가 커지게 되면 이 상관계수를 한 눈에 파악하기에 불편해짐 따라서 이를 히트맵으로 나타내어 파악하기 쉽게 만듦

```python
corr=df[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('correlation coefficient')
# plt.savefig('tip_heatmap.png')    # 주의: 반드시 파일로 저장하는 plt.savefig()를 plt.show() 전에 해야 함
plt.show()
```

```python
corr=df[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('correlation coefficient')
plt.show()
```

해석예

- total_bill과 tip은 양의 상관관계를 보임 (결제 금액이 많을수록 팁도 큼)  

- size와 total_bill도 자연스러운 관계 (사람 많을수록 결제금액 증가)  

- 흡연자 vs 비흡연자의 팁 분포 차이를 시각적으로 파악 가능
