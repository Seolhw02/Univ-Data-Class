## 중요

- train_test_split() 은 **공부용**과 **시험용**으로 데이터를 나누는 함수
- 시험(test) 데이터는 모델의 진짜 실력을 측정하는 데 사용
- test_size (또는 train_size) 를 조정하여 훈련/시험 비율을 조절
