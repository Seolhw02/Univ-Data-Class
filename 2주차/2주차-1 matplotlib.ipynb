### plt.xlabel(), plt.ylabel(), plt.title()  한줄씩 추가하며 실행시켜보세요!

```python
plt.plot([1, 2, 3, 4], 'ro') #red o 
plt.title('Example',size=15)   
plt.xlabel('x numbers') #size
plt.ylabel('y numbers') #size
plt.show()
```

```python
plt.plot([1, 2, 3, 4], 'ro')
plt.title('Example',size=15)
plt.xlabel('x numbers', size=12)
plt.ylabel('y numbers', size=12)
plt.show()
```
