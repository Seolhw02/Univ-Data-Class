### Covid19 데이터 
#### 날짜별 한국 확진자 수, 미국 확진자 수     *주의:리스트 자료구조

위의 셀 코드 다시 써보기

```python
dates=['1/22', '2/1', '2/15', '2/28', '3/10', '3/23']
kor_cases=[1, 15, 50, 200, 500, 600]
us_cases=[5, 20, 70, 500, 900, 1500]
```

### 날짜별(x축) 한국 확진자 수 (y축) 를 plot 할때 plot() 함수  옵션 연습 

```python
plt.figure(figsize=(12,8))
plt.plot(dates,kor_cases, c='green',ls='dashed',lw=3, marker='o')
plt.show()
```

### 한글 폰트와 - 깨짐 해결코드

```python
plt.rc('font', family='Malgun Gothic') 
plt.rc('axes', unicode_minus = False)
```

### 그래프 장식 연습

```python
plt.figure(figsize=(16,9))
plt.plot(dates, kor_cases,lw=3, label='한국')
plt.plot(dates, us_cases, label='US')
plt.legend(fontsize=25)
plt.xlabel('날짜',size=20)
plt.ylabel('# of confirmed cases',size=20)
plt.title('COVID-19',size=30)
plt.text('2/15', 200, '급격히 증가함',size=20)
plt.grid()
plt.xticks(size=20)
plt.yticks(size=20)
plt.show()
```

위의 셀 코드 다시 써보기

```python
plt.figure(figsize=(16,9))
plt.plot(dates, kor_cases,lw=3, label='한국')
plt.plot(dates, us_cases, label='US')
plt.legend(fontsize=25)
plt.xlabel('날짜',size=20)
plt.ylabel('# of confirmed cases',size=20)
plt.title('COVID-19',size=30)
plt.text('2/15', 200, '급격히 증가함',size=20)
plt.grid()
plt.xticks(size=20)
plt.yticks(size=20)
plt.show()
```
